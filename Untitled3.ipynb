```python
# ============================================
# Analyse du chômage au Maroc (1999-2025)
# Source : HCP - Enquête nationale sur l'emploi
# ============================================

import pandas as pd
import matplotlib.pyplot as plt

# 1. Importer le fichier CSV
from google.colab import files

uploaded = files.upload()

# 2. Lire le fichier CSV
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

# 3. Afficher les premières lignes pour vérifier les données
print("Données importées :")
display(df.head())

# 4. Trier les années dans l'ordre chronologique
df = df.sort_values("annee")

# 5. Créer le graphique du taux de chômage global
plt.figure(figsize=(12, 6))

plt.plot(
    df["annee"],
    df["ensemble"],
    marker="o",
    linewidth=2
)

# 6. Ajouter le titre et les noms des axes
plt.title(
    "Évolution du taux de chômage au Maroc (1999-2025)",
    fontsize=15
)

plt.xlabel("Année", fontsize=12)
plt.ylabel("Taux de chômage (%)", fontsize=12)

# 7. Ajouter une grille
plt.grid(True, alpha=0.3)

# 8. Afficher le graphique
plt.show()
```
